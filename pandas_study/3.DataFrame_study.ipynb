#### 一、创建DataFrame

```python
import pandas as pd
import numpy as np
```

##### 1.通过series创建DataFrame


```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({"第一列":s1, "第二列":s2})
print(df)
type(df['第一列'])
```

##### 2.通过字典创建

```python
df = pd.DataFrame(
    {
        # "id":[1,2,3,4,5],
        "name":["tom", "jack", "alice", "bob", "charlie"],
        "age":[10, 20, 30, 40, 50],
        "score":[60.1, 70.2, 80.3, 90.4, 100]
    }, index=[1,2,3,4,5], columns=["name","score","age"] # 设置列的顺序
)
df
```

#### 二、DataFrame的属性


```python
print("行索引：")
print(df.index)
print("列索引：")
print(df.columns)
print("值：")
print(df.values)
```

```python
print("维度：")
print(df.ndim)
print("形状:")
print(df.shape)
print("元素个数：")
print(df.size)
print("数据类型")
print(df.dtypes)
```

##### 行列转置

```python
print(df.T)
print(df.T.index)
print(df.T.columns)
```

##### 获取元素


```python
#获取某行
print(df.loc[4])
print(df.iloc[4])
```

```python
#获取某列
print(df.loc[:, 'name'])
print(df.iloc[:, 0])
```

```python
#单个元素
print(df.at[1, 'name'])
print(df.iat[1, 1])
```

```python
#获取单列数据
print(df['name'])
print(type(df['name']))
print(df.name)
print(type(df.name))
print(df[['name']])
print(type(df[['name']]))  #DataFrame
```

```python
# 获取多列数据
print(df[['name', 'age']])
print(type(df[['name', 'age']]))  #DataFrame
```

```python
# 产看部分数据
print(df.head())
print(df.tail())
```

##### 布尔索引筛选数据

```python
df.score > 70
```

```python
df[df.score > 70]
```

```python
df[(df.score > 70) & (df.age > 30)]
df[(df['score'] > 70) & (df['age'] > 30)]
```

##### 随机抽样

```python
df.sample(2)
```

#### 三、DataFrame的常用方法


##### 1.查看元素是否在参数集合中

```python
 print(df.isin(['jack', 20]))
```

##### 2.查看元素是否是缺失值

```python
print(df.isna())
```

##### 3.统计信息

```python
print(df["score"].sum())  # 求和
print(df.score.max())  # 最大值
print(df.age.min())  # 最小值
print(df.score.mean())  # 平均值
print(df.score.median())  # 中位数
print(df.age.mode())  # 众数
print(df.score.std())   # 标准差
print(df.score.var())   # 方差
print(df.score.quantile(0.25)) # 25%分位数
print(df.score.quantile(0.75)) # 75%分位数
```

```python
print(df.describe())
```

##### 4.缺失值处理

```python
print(df.count())  # 非缺失值的个数
```

```python
print(df.value_counts())  # 每个值的个数
```

##### 5.重复值处理

```python
print(df.drop_duplicates())  # 删除重复值
```

```python
print(df.duplicated(subset=['name']))  # 查看重复值
```

##### 6.随机抽样

```python
df.sample(2)
```

##### 7.替换值

```python
print(df.replace(10, 30))
```

##### 8.累计求和

```python
df.cumsum(axis=0)  # 按列累计求和
```

##### 9.排序

```python
print(df.sort_index(ascending=False)) # 按行索引降序排序
```

```python
print(df.sort_values(by=['score', 'age'], ascending=[True,False])) # 按score列升序，age列降序排序
```

##### 10.取最大值和最小值

```python
df.nlargest(2, columns=['score', 'age'])  # 取score和age两列中最大的2行
```

```python
df.nsmallest(2, columns=['score', 'age'])  # 取score和age两列中最小的2行
```

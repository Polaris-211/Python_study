#### 五、索引与切片

```python
import numpy as np
```

##### 1.向量索引与切片

```python
arr = np.random.randint(1, 100, 20)
print(arr)
```

```python
print(arr[0])
print(arr[:]) #获取全部元素
print(arr[0:14:3])
print(arr[arr > 10])  #布尔索引
print(arr[slice(2, 18, 3)])
```

##### 2.矩阵索引与切片

```python
arr = np.random.randint(1, 100, (4, 8))
print(arr)
```

```python
print(arr[1, 3])
print(arr[1, :])  #arr[始行:终行, 始列:终列]
print(arr[arr > 50])
print(arr[2][arr[2] > 50])
print(arr[:, 3][arr[:, 3] > 50])
```

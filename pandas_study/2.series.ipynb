#### 一、series的创建

```python
import pandas as pd
s = pd.Series([10,2,3,4,5])
print(s)
```

##### 1.自定义索引

```python
s = pd.Series([10,2,3,4,5], index = ['a','b','c','d','e'])
print(s)
```

##### 2.定义name

```python
s = pd.Series([10,2,3,4,5], index = ['a','b','c','d','e'], name = 's1')
print(s)
```

##### 3.通过字典创建

```python
s = pd.Series({'a':10, 'b':2, 'c':3, 'd':4, 'e':5})
print(s)
s1 = pd.Series(s, index = ['a', 'c'])
print(s1)
```

#### 二、series属性


##### 1.index：索引

```python
print(s.index)
```

##### 2.values：值

```python
print(s.values)
```

##### 3.dtype：元素的类型

```python
print(s.dtype)
```

##### 4.shape：形状

```python
print(s.shape)
```

##### 5.ndim：维度数

```python
print(s.ndim)
```

##### 6.size：元素个数

```python
print(s.size)
```

##### 7.name：series的名称

```python
print(s.name)
```

##### 8.loc[] 显式索引：按标签索引或切片

```python
print(s.loc['a'])
print(s.loc['a':'c'])
```

##### 9.iloc[] 隐式索引：按位置索引或切片

```python
print(s.iloc[0])
print(s.iloc[0:2])
```

##### 10.at[] 使用标签访问单个元素

```python
print(s.at['a'])
```

##### 11.iat[] 使用位置访问单个元素

```python
print(s.iat[0])
```

#### 三、访问数据

##### 1.直接访问

```python
# print(s[0])
print(s['a'])
```

##### 2.布尔索引

```python
print(s[s < 5])
```

##### 3.表头

```python
s['f'] = 6
print(s.head())
print(s.head(2))  #默认显示前5个元素
print(s.tail(2))  #默认显示后5个元素
```

#### 三、常用方法

```python
import numpy as np


s = pd.Series([10, 2, np.nan, None, 3, 4, 5], index = ['a','b','c','d','e','f','g'])
print(s)
```

##### 1.查看所有描述性信息

```python
s.describe()
```

##### 2.获取元素个数 （忽略缺失值）

```python
print(s.count())
```

##### 3.获取索引

```python
print(s.index)  #属性
print(s.keys())  #方法
```

##### 4.判断缺失值

```python
print(s.isna())
```

##### 5. 判断是否在在参数集合中

```python
print(s.isin([4, 5,]))
```

##### 6.统计方法（忽略缺失值）

```python
s['h'] = 4
print(s.mean())    #均值
print(s.std())     #标准差
print(s.var())     #方差
print(s.min())     #最小值
print(s.max())     #最大值
print(s.mode())    #众数
print(s.value_counts())
print(s.median())  #中位数
print(s.quantile(0.25))  #四分位数
print(s.quantile(0.75))  #四分位数
```

##### 7.排序


```python
print(s.sort_values()) #按值排序
print(s.sort_index())  #按索引排序
```

##### 8.去重

```python
print(s.drop_duplicates())
print(s.unique())
print(s.nunique())  #去重后的元素个数
```

#### 一、缺失值处理

```python
import pandas as pd
import numpy as np
s = pd.Series([1, 2, np.nan, None, pd.NA])
df = pd.DataFrame([[1, pd.NA, 2], [2, 3, 5], [None, 4, 6]], columns=['第一列', '第二列', '第三列'])
print(s)
print(df)
```

##### 1.查看是否有缺失值

```python
print(s.isna())
print(s.isnull())
print(s.isna().sum())
print(df.isna())
print(df.isnull())
print(df.isna().sum(axis=0))  #
```

##### 2.剔除缺失值

```python
print(s.dropna())
print(df.dropna())  #默认剔除包含缺失值的行
print(df.dropna(how='all'))  #剔除所有缺失值的行
print(df.dropna(thresh = 2))  #如果至少有n个值不是缺失值，就保留该行
print(df.dropna(axis=1))  #剔除包含缺失值的列
print(df.dropna(subset=['第一列']))  #剔除第一列包含缺失值的行
```

##### 3.填充缺失值


```python
print(df.fillna({'第一列': 0, '第二列': 1, '第三列': 2}))  #使用字典填充固定缺失值
print(df.fillna(df[['第一列', '第二列']].mean()))  #使用平均值填充
print(df.ffill())  #前向填充缺失值
print(df.bfill())  #后向填充缺失值
```

#### 二、重复值处理

```python
data = {
    "name": ['alice', 'alice', 'bob', 'alice', 'jack', 'bob'],
    "age": [26, 25, 30, 25, 35, 30],
    'city': ['NY', 'NY', 'LA', 'NY', 'SF', 'LA'],
    'gender': ['male', 'female', 'male', 'female', 'male', 'female']
}
df = pd.DataFrame(data)
df
```

##### 1.查看是否有重复值

```python
print(df.duplicated())
print(df.drop_duplicates(subset=['name']))  #根据指定列剔除重复值
print(df.drop_duplicates(subset=['name'], keep='last'))  #保留最后一个重复值
```

#### 三、数据类型转换

```python
df['age'] = df['age'].astype('int16')
df.dtypes
```

```python
df['city'] = df['city'].astype('category')
print(df.city)
df.dtypes
```

```python
df['is_male'] = df['gender'].map({'male':True, 'female':False})
df.is_male
```

#### 五、分组聚合

##### df.groupby('分组的字段')['聚合的字段'].聚合函数()

```python
import pandas as pd
df = pd.read_csv('employees.csv')
print(df.department_id.isna().sum())
df = df.dropna(subset=['department_id'])
print(df.department_id.isna().sum())
df.department_id = df.department_id.astype('int64')
df
```

##### 计算不同部门的平均工资


```python
df.groupby('department_id').groups  #查看分组结果
df.groupby('department_id').get_group(10)  #查看具体某个分组信息
df2 = df.groupby('department_id')[['salary']].mean()  #计算每个部门的平均工资
```

```python
df2['salary'] = df2['salary'].round(2)
df2 = df2.reset_index()
```

```python
df2.sort_values('salary', ascending=False)
```

##### 计算不同部门不同岗位的人平均薪资

```python
df2 = df.groupby(['department_id', 'job_id'])[['salary']].mean()
df2 = df2.reset_index()
df2['salary'] = df2['salary'].round(1)  #保留1位小数
df2.sort_values('salary', ascending=False)
```
